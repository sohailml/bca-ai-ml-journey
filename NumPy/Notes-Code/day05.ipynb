### Condition array
- condition array is similar to if else just like in python.
- syntax is `np.where(numbers>5,numbers*4,numbers)` so first **numbers>5** is a condition so if this condition will true so execuate second **numbers*4** else execuate numbers.
- it will print [1 2 3 4 5 24 28 32 36 40] so 1 to 5 is less then 5 so it does not satisfy condition thats why its print without no change and after 6 to 10 these are satisfying the condition thats why it multiply by 4.

```python
condition_arr =np.where(numbers>5,numbers*4,numbers
                        )
print("condition_array:",condition_arr)
```

### Adding and Removing Data
- First combine/merging array or concatenate so we can merge two or more array in one by `np.concatenate((arr1,arr2))` this will take arguments in Tuple

```python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
combine = np.concatenate((arr1,arr2))
print(combine)
```

### Array Compatiblity
- basically it is to find array are compatible or not in shape,dimension to performe task like adding,subtracting or others

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
print("compatiblity shape:",a.shape == b.shape)
```

#### Adding array by row or coloumn
- **np.vstack** it is use to add new row. it take arguments in tuple
- **np.hstack** it is use to add new coloumn. it take argumnets in tuple

```python
original = np.array([[1,2],[3,4]])
new_row = np.array([[5,6]])
with_new_row = np.vstack((original,new_row))

print("original:",original)
print("with_new_row:",with_new_row)

new_col =np.array([[7],[8]])
with_new_col = np.hstack((original,new_col))
print("with_new_col:",with_new_col)
```

- **np.delete()** from this we can delete an element from an array it take two parameters first array name and second index of that element

```python
arr = np.array([1,2,3,4,5,6])
deleted = np.delete(arr,3)
print("array after deletion:",deleted)
```
