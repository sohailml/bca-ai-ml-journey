## properties of Array
There ara many properties of an array
`arr = np.array([[1,2,3],[4,5,6]])`

- 1. Shape - we can find shape of an array means how many row and coloumn there are so this is the syntax **arr.shape**
- 2. Dimension - we can find dimension of an array whether it is 2d or 3d or ndimension so there is the syntax **arr.ndim**
- 3. Size - basically it will give number of all the elements present in array so here is syntax **arr.size**
- 4. Data Type - so it give the type of array we generally perfer similar data type means if we are using int so we use all element int **arr.dtype**

For now these are some properties i learn more as i move next.

```python
import numpy as np
```

```python
arr = np.array([[1,2,3],
[4,5,6]])
print("Shape:",arr.shape)
print("Dimension:",arr.ndim)
print("Size:",arr.size)
print("Dtype:",arr.dtype)
```

## Array Reshaping
- so we can reshape an array in any matrix means in row and colounm `arr = np.arange(12)` so we can reshape into (3,4) matrix or any here is syntax `arr.reshape(3,4)`

- and we can also flatten an array mean in one row by `reshaped.flatten()` 
- revel is same like flatten it also give array in one row but flatten give copy and revel give view original here is syntax `reshaped.revel()`
- Transpose so i study this in my 12th class basically in transpose row and column change means if a matrix is (3 x 4) so its transpose matrix will be of (4 x 3) here is syntax for tranpose of an array `reshaped.T`

```python
arr = np.arange(12)
print("original array:",arr)

reshaped = arr.reshape(3,4)
print("reshaped array:",reshaped)

flattened = reshaped.flatten()
print("flattened array:",flattened)

reveled = reshaped.ravel() # it give view insted of copy
print("reveled array:",reveled)

transposed = reshaped.T
print("trnsposed array:",transposed)
```
