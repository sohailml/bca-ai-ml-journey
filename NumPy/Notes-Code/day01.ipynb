### Creating array from list
- **np.array()** se array bante hai sare ke sare and is np.array() ko ek hi arguments chahiye hote hai **list**

```python
arr_1d = np.array([1,2,3,4,5])
print("1d array:",arr_1d)
```

#### 2d
`arr_2d = np.array([1,2,3],[4,5,6])`
isme error aa jayega kyunki np.array() sirf ek arguments leta hai or hamne 2 de diye.  
so ham [[1,2,3],[4,5,6]] aise use karenge.

```python
arr_2d = np.array([[1,2,3],[4,5,6]])
print("2d array:",arr_2d)
```

### List vs numpy array
- python multiplication bahut easy hai bas ye  multiply karke 2 bar print kar deta hai
- numpy array multiplication mai kisi constant se multiply karte hai jaise **np_array*2** to ye one by one element mai multiply hota hai and isko **scaler se multiply** karna bhi bolte hai

```python
py_list = [1,2,3]
print("python list multiplication",py_list*2)

np_array = np.array([1,2,3])
print("python array multiplication",np_array*2)
```

#### .arange()
- so **np.arange()** basically ek numpy ka function hai isse ham ek bar mai badi range ka array bana sakte hai
- start = Starting value, stop = Ending limit (excluded), step = Difference between values. ye iske parameters hote hai

```python
import time
start = time.time()
py_list = [i*2 for i in range(1000000)]
print("\n list operation time:",time.time() - start)

start = time.time()
np_array = np.arange(1000000) * 2
print("\n Numpy operation time:",time.time() - start)
```
