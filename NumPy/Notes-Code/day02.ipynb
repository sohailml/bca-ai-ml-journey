## Creating array from scratch

```python
import numpy as np
```

```python
ones = np.ones((3,4))
print(ones)

zero = np.zeros((2,3))
print(zero)

full = np.full((3,3),8)
print(full)

random = np.random.random((5,5))
print(random)

sequence = np.arange(0,10,2)
print(sequence)
```

## Vector,Matrix and Tensor

```python
vector = np.array([1,2,3])
print("Vector:",vector)

matrix = np.array([[1,2,3],
                   [4,5,6]])
print("Matrix:",matrix)

tensor =np.array([[[1,2],[3,4]],
                  [[5,6],[7,8]]])
print("Tensor:",tensor)
```
