### Operation with Business examples

```python
import numpy as np
import matplotlib.pyplot as plt
```

- sales_data[:,1:] means all rows and then givin range of column

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("=== Zomato sales analysis ===")
print("\n sales data shape:",sales_data.shape)
print("\n sales data size:",sales_data.size)
print("\n sample data for 1st 3 restaurant:",sales_data[0:3])
print("\n sample data for restaurant only sales:",sales_data[:,1:])
```

`np.sum()` by this we can find sum of data it take two parameter name of array with slicing and axis, axis = 0 so top to bottome will be added and axis = 1 so left to right.

```python
# total sales per year
yeary_sales = np.sum(sales_data[:,1:],axis=0)
print(yeary_sales)
total_sales = np.sum(yeary_sales,axis=0)
print(total_sales)
```

`np.min()` is to use for finding minimum it take two parameters same as np.sum

```python
# minimum sales per restaurent
min_sales = np.min(sales_data[:,1:],axis=1)
print(min_sales)
```

`np.max()` is to use maximum

```python
# max sales per year
max_sales = np.max(sales_data[:,1:],axis=0)
print(max_sales)
```

`np.mean()` is use to find average. it also take two parameters.

```python
#avg sales per restaurent
avg_sales = np.mean(sales_data[:,1:],axis=1)
print(avg_sales)
```

cumulative sum is basically previous total + current it can be done by `np.cumsum()`

```python
# Cumulative Sales= Previous Total + Current Sales
cumsum = np.cumsum(sales_data[:,1:],axis=1)
print(cumsum)
```

#### Small overview of matplotlib
- `plt.figure(figsize=(10,6))` we want figure first figsize is size of figure/graph it take parameters in tuple
- `plt.plot(np.mean(cumsum,axis=0))` so it used to ploting on figure so what it plot mean of cumsum.
- `plt.title("")` used for giving title of graph
- `plt.xlabel("")` used for giving title for x axis
- `plt.ylabel("")` used for giving title for y axis
- `plt.grid(True)` for girds
- Last `plt.show()` for showing graph.

```python
plt.figure(figsize=(10,6))
plt.plot(np.mean(cumsum,axis=0))
plt.title("Average cumulative sales per year across all resturant")
plt.xlabel("years")
plt.ylabel("sales")
plt.grid(True)
plt.show()
```
