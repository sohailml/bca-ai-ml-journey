## 🎯 Lecture 19: Gradient Descent for Linear Regression (Derivatives Nikalna)

### 🔄 Ab Tak Kya Tha (3 Pieces)

| Piece | Kya Tha |
|---|---|
| **1. Model** | f(x) = wx + b |
| **2. Cost Function** | J(w,b) = (1/2m) Σ(f(x) - y)² |
| **3. Gradient Descent** | w := w - α·∂J/∂w, b := b - α·∂J/∂b |

### 🔥 Aaj Ka Kaam: **Ab Teen Pieces Ko Jodo!**

**Problem:** Gradient descent mein **∂J/∂w** aur **∂J/∂b** chahiye the. Ye **kaise nikalte hain**? Ye video wahi batata hai!

---

### 📐 Derivative Formulas (Ye Yaad Rakhna — Code Mein Use Honge!)

$$\frac{\partial J}{\partial w} = \frac{1}{m} \sum_{i=1}^{m} \left( f(x^{(i)}) - y^{(i)} \right) \cdot x^{(i)}$$

$$\frac{\partial J}{\partial b} = \frac{1}{m} \sum_{i=1}^{m} \left( f(x^{(i)}) - y^{(i)} \right)$$

> ⚠️ **Bhai, ghabrana mat!** Ye formulas **calculus se aate hain**, lekin **tujhe calculus nahi seekhni**. Bas **kaise use karna hai** samajh le. Course mein bola: **"Calculus skip kar sakte ho, sab kaam karega!"**

---

### 🧠 Formula Ka INTUITION (Sabse Important Part!)

#### 🔍 Common Part: **Error Term**

$$\text{Error}_i = f(x^{(i)}) - y^{(i)} = (\text{Prediction} - \text{Actual})$$

- **Positive error** → Prediction **zyada** hai (upar)
- **Negative error** → Prediction **kam** hai (neeche)

#### 📌 ∂J/∂w Ka Formula — Kyun `× x` Hai?

$$\frac{\partial J}{\partial w} = \frac{1}{m} \sum \underbrace{(\text{Error}_i)}_{\text{pred - actual}} \cdot \underbrace{x^{(i)}}_{\text{input}}$$

**Intuition:** w ka model mein role: **f(x) = w·x + b** — w ko **x se multiply** kiya jaata hai. Agar **x bada** → **w ka error pe bada impact**. Isliye w ke derivative mein **x se multiply** karte hain.

> 🧠 **Analogy:** **x** = "Kitna bada lever" (jitna bada x, utna zyada w ka asar). **Error × x** = "w ki galti mein kitna hissa hai."

#### 📌 ∂J/∂b Ka Formula — Kyun `× x` NAHI Hai?

$$\frac{\partial J}{\partial b} = \frac{1}{m} \sum (\text{Error}_i)$$

**Intuition:** b ka model mein role: **f(x) = wx + b** — b ko **x se multiply NAHI** kiya — bas **direct add** kiya. Isliye b ke derivative mein **sirf error** hai, x nahi.

---

### 🆕 Complete Mathematical Derivations (Chain Rule + Power Rule)

> 🧠 **Bhai, ye section calculus ke liye hai. Agar tu calculus skip karna chahta hai — koi baat nahi, formulas yaad rakh, kaam ho jayega! Lekin agar curiosity hai — ye padh, sab clear ho jayega.**

#### 🔧 Setup

Cost: $J(w,b) = \frac{1}{2m} \sum_{i=1}^{m} \left( f_{w,b}(x^{(i)}) - y^{(i)} \right)^2$

Model: $f_{w,b}(x^{(i)}) = w \cdot x^{(i)} + b$

**Substitution** (calculation easy karne ke liye):
$$\text{Let } e_i = f_{w,b}(x^{(i)}) - y^{(i)} = w \cdot x^{(i)} + b - y^{(i)}$$

**Toh cost function:**
$$J(w,b) = \frac{1}{2m} \sum_{i=1}^{m} e_i^2$$

---

#### 🎯 Derivation of ∂J/∂w (Step-by-Step)

**Step 1 — Linearity of derivative** (sum ke andar derivative le sakte hain):
$$\frac{\partial J}{\partial w} = \frac{1}{2m} \sum_{i=1}^{m} \frac{\partial}{\partial w}\left( e_i^2 \right)$$

**Step 2 — Chain Rule** (Power Rule + Chain Rule):
> **Power Rule:** $\frac{d}{dx}(u^n) = n \cdot u^{n-1} \cdot \frac{du}{dx}$ — for n=2: $\frac{d}{dx}(u^2) = 2u \cdot \frac{du}{dx}$

$$\frac{\partial}{\partial w}(e_i^2) = 2 \cdot e_i \cdot \frac{\partial e_i}{\partial w}$$

**Step 3 — Substitute back:**
$$\frac{\partial J}{\partial w} = \frac{1}{2m} \sum_{i=1}^{m} 2 \cdot e_i \cdot \frac{\partial e_i}{\partial w}$$

**Step 4 — Simplify** (2 cancel ho gaya):
$$\frac{\partial J}{\partial w} = \frac{1}{m} \sum_{i=1}^{m} e_i \cdot \frac{\partial e_i}{\partial w}$$

**Step 5 — Compute $\partial e_i/\partial w$:**
$$e_i = w \cdot x^{(i)} + b - y^{(i)}$$
$$\frac{\partial e_i}{\partial w} = x^{(i)} \quad \checkmark$$
*(Kyunki $\frac{\partial}{\partial w}(w \cdot x^{(i)}) = x^{(i)}$; b aur y^(i) constants hain — derivative 0.)*

**Step 6 — Final result:**
$$\boxed{\frac{\partial J}{\partial w} = \frac{1}{m} \sum_{i=1}^{m} \left( f_{w,b}(x^{(i)}) - y^{(i)} \right) \cdot x^{(i)}} \quad \checkmark$$

---

#### 🎯 Derivation of ∂J/∂b (Step-by-Step)

**Steps 1-4 — Same** (upar wale):
$$\frac{\partial J}{\partial b} = \frac{1}{m} \sum_{i=1}^{m} e_i \cdot \frac{\partial e_i}{\partial b}$$

**Step 5 — Compute $\partial e_i/\partial b$:**
$$\frac{\partial e_i}{\partial b} = 1 \quad \checkmark$$
*(Kyunki $\frac{\partial}{\partial b}(w \cdot x^{(i)}) = 0$; $\frac{\partial}{\partial b}(b) = 1$; $\frac{\partial}{\partial b}(y^{(i)}) = 0$.)*

**Step 6 — Final result:**
$$\boxed{\frac{\partial J}{\partial b} = \frac{1}{m} \sum_{i=1}^{m} \left( f_{w,b}(x^{(i)}) - y^{(i)} \right)} \quad \checkmark$$

---

#### 📌 Dono Derivations — Ek Table Mein

| Derivative | Formula | Key Step |
|---|---|---|
| **∂J/∂w** | (1/m) Σ (Error_i · x^(i)) | $\partial e_i/\partial w = x^{(i)}$ |
| **∂J/∂b** | (1/m) Σ (Error_i) | $\partial e_i/\partial b = 1$ |

> 🧠 **Yaad Rakhne Ki Trick:** Sirf **ek difference** hai — w wale mein **× x^(i)** hai, b wale mein **nahi**. Kyunki w multiply hota hai x se, b nahi.

---

### 🆕 Why "1/2" in Cost Function? (Aksar Poocha Jaata Hai)

#### ❓ Mystery
Cost function: $J(w,b) = \frac{1}{2m}\sum(f(x^{(i)}) - y^{(i)})^2$

**Question:** 2 kyun hai? 1/m bhi kaafi tha!

#### 🎯 Answer: Neat Derivatives

**Upar derivation Step 2 mein dekha:**
$$\frac{\partial}{\partial w}(e_i^2) = 2 \cdot e_i \cdot \frac{\partial e_i}{\partial w}$$

**Power Rule se "2" neeche aaya!** Ab dekho kya hota hai:

| Scenario | Cost Function | ∂J/∂w |
|---|---|---|
| **Without 1/2** | $J' = \frac{1}{m} \sum e_i^2$ | $\frac{\partial J'}{\partial w} = \frac{2}{m} \sum e_i \cdot x^{(i)}$ ← extra 2! |
| **With 1/2** | $J = \frac{1}{2m} \sum e_i^2$ | $\frac{\partial J}{\partial w} = \frac{1}{m} \sum e_i \cdot x^{(i)}$ ← clean! ✓ |

**Dekha? 1/2 se derivative clean ho jaata hai!** Ek akela 2 cancel ho jaata hai. Isse math **neat** lagti hai.

#### 🤔 "1/2 se minimum change ho jaata hai?"

**Nahi!** Minimizing $J$ vs minimizing $\frac{1}{2}J$ — **same (w, b) pe minimum** hoga.

**Reason:** Positive constant se multiply karne se **minimum ka location nahi badalta**, sirf **value scale** hoti hai.
$$\arg\min_{w,b} J(w,b) = \arg\min_{w,b} \frac{1}{2}J(w,b) \quad \checkmark$$

**Gradient descent ko farak padta hai?** Nahi! Kyunki update rule hai:
$$w := w - \alpha \cdot \frac{\partial J}{\partial w}$$
Agar 1/2 na hota, toh ∂J/∂w mein **2× zyada** aata. Hum **α ko half** karke same result laa sakte the. **Matlab 1/2 + α ka combination same kaam karta hai.** Isliye 1/2 convention hai — math clean rakhne ke liye.

#### 🧮 Numerical Verification

**Data:** x=[1,2], y=[300,500], w=100, b=100

| Version | Formula | Cost Value |
|---|---|---|
| **With 1/2** | $\frac{1}{4}[(-100)^2 + (-200)^2]$ | $\frac{50000}{4} = 12500$ |
| **Without 1/2** | $\frac{1}{2}[(-100)^2 + (-200)^2]$ | $\frac{50000}{2} = 25000$ |

**Dono ka minimum same (w=200, b=100).** Bas value factor 2 se different. ✓

> 💡 **Ek Line Mein:** "1/2 sirf isliye hai ki derivative clean ho — mathematically koi farak nahi padta ki 1/2 ho ya na ho."

- f(x⁽¹⁾) = 100(2) + 100 = **300** → Error = 300 - 500 = **-200**

**Derivatives:**
$$\frac{\partial J}{\partial w} = \frac{1}{2}[(-100)(1) + (-200)(2)] = \frac{-500}{2} = \boxed{-250}$$
$$\frac{\partial J}{\partial b} = \frac{1}{2}[(-100) + (-200)] = \boxed{-150}$$

**Update (α = 0.01):**

> 🧠 **Dekh:** Negative derivative → **w aur b badhe** (upar gaye). Ye **sahi** hai — predictions **kam** thi (actual se neeche), toh **upar** jaana chahiye!

#### 📌 Example 3: Starting from Zero (w=0, b=0) — Initial State

**Predictions:**
- f(1) = 0 → Error = **-300**
- f(2) = 0 → Error = **-500**

**Derivatives:**
$$\frac{\partial J}{\partial w} = \frac{1}{2}[(-300)(1) + (-500)(2)] = \frac{-1300}{2} = \boxed{-650}$$
$$\frac{\partial J}{\partial b} = \frac{1}{2}[-300 + (-500)] = \boxed{-400}$$

**Cost Before Update:**
$$J = \frac{1}{4}[(-300)^2 + (-500)^2] = \frac{90000 + 250000}{4} = \boxed{85000}$$

**Update (α = 0.01):**
- w := 0 - 0.01(-650) = **6.5**
- b := 0 - 0.01(-400) = **4.0**

**New Predictions:** f(1)=10.5, f(2)=17 → Errors: -289.5, -483
**New Cost:** $\frac{1}{4}[289.5^2 + 483^2] = \frac{83810.25 + 233289}{4} = \boxed{79274.81}$

> 🧠 **Dekh:** Cost **85000 → 79274.81** — **decrease** hui! ✅ Gradient descent sahi direction mein chal raha hai. **α too large nahi hai** (converge ho raha hai).

**Python Check:**
```python
w, b, alpha = 0, 0, 0.01
dw = ((w*x[0]+b-y[0])*x[0] + (w*x[1]+b-y[1])*x[1]) / 2   # = -650
db = ((w*x[0]+b-y[0]) + (w*x[1]+b-y[1])) / 2              # = -400
w_new = w - alpha * dw   # = 6.5
b_new = b - alpha * db   # = 4.0
print(w_new, b_new)   # Output: 6.5 4.0  ✓
```

> 🔥 **Warning Sign (Divergence):** Agar α = 0.8 hota (bahut bada):
> - w := 0 - 0.8(-650) = **520**
> - b := 0 - 0.8(-400) = **320**
> - New cost ≈ **257800** ← **BADH gaya!** (Diverging! ❌)

---

### ⚠️ SIMULTANEOUS UPDATE (Phir Se Yaad Karo!)

```python
# ✅ CORRECT (simultaneous)
temp_w = w - α·(∂J/∂w)   ← purane w se calculate
temp_b = b - α·(∂J/∂b)   ← purane b se calculate

w = temp_w   ← dono update ek saath
b = temp_b
```

> 🧠 **Yaad Rakho:** **Dono derivatives purane w, b se** nikalo. **Phir dono update karo.**

---

### 🥣 Convex Function Ka Magic (BIG REASON To Celebrate!)

**Earlier concern:** Gradient descent **local minimum** mein phas sakta hai (hill analogy — valley 1 ya valley 2).

**BUT — Linear Regression Ke Liye:**

> ✅ **Squared Error Cost Function = CONVEX (bowl shape)**
> 
> ✅ **Sirf EK global minimum hai — koi local minima nahi!**
> 
> ✅ **Gradient descent HAMESHA global minimum tak pahunchega** (agar α sahi ho)

**Matlab:** Multiple valleys **nahi** hain. Ek hi **bowl** hai. Gradient descent **kahan se bhi start karo** — bowl ke **bottom** pe hi jayega! 🥣🎯

> 💡 **Formal:** **Convex function** = Bowl-shaped function jisme **sirf ek global minimum** ho, koi local minimum nahi.

> 🧠 **Isliye ye tension-free hai!** Neural networks mein multiple local minima hote hain (problem), lekin **linear regression mein nahi** (because convex). **Ye linear regression ko simple aur reliable banata hai.**

### 📦 "BATCH" Gradient Descent Kya Hai?

**Naam ka matlab:**

> **Batch Gradient Descent** = Har step pe **saare training examples** dekhta hai (entire batch).

**Derivative formula mein:**
$$\frac{\partial J}{\partial w} = \frac{1}{m} \sum_{i=1}^{m} \ldots$$

- **Σ from i=1 to m** — **saare m examples** ka sum
- Matlab: **saare examples ka use** hua is step mein

### 📊 Batch vs Other Types

| Type | Har Step Mein Kya Dekhta | Example |
|---|---|---|
| **Batch GD** | **Saare m examples** | Linear regression (is course mein) |
| **Mini-batch GD** | **Chhota subset** (e.g., 32 examples) | Deep learning mein common |
| **Stochastic GD** | **Ek hi example** at a time | Quick updates |

> 💡 **Is course mein:** Linear regression ke liye **Batch Gradient Descent** use karenge.
> 
> **Fun fact:** DeepLearning.AI ka newsletter **"The Batch"** isi concept se named hai! 📰

## 🧪 Lab: Gradient Descent for Linear Regression (Line-by-Line)

> 🧠 **Bhai, lab ka code ab samjhaunga — har line ka matlab. Ye lab ke 3 functions hain: `compute_cost`, `compute_gradient`, `gradient_descent`.**

### 📊 Data (Wahi 2 Points)

```python
x_train = np.array([1.0, 2.0])   # Size (1000 sqft)
y_train = np.array([300.0, 500.0])  # Price (1000s of $)
```

---

### 💻 Function 1: `compute_cost` (Pichle Lab Se)

```python
def compute_cost(x, y, w, b):
    m = x.shape[0]
    cost = 0
    for i in range(m):
        f_wb = w * x[i] + b
        cost = cost + (f_wb - y[i])**2
    total_cost = 1 / (2 * m) * cost
    return total_cost
```

**Kya karta hai:** J(w,b) = (1/2m) Σ(prediction - actual)² — **wahi purana formula**

---

### 💻 Function 2: `compute_gradient` (NAYA — Dhyan Se!)

> 🔥 **Ye aaj ka core function hai. Isko line-by-line samjho!**

```python
def compute_gradient(x, y, w, b):
    m = x.shape[0]           # Step 1: Kitne examples
    dj_dw = 0                # Step 2: ∂J/∂w ka sum start (0)
    dj_db = 0                # Step 3: ∂J/∂b ka sum start (0)
    
    for i in range(m):       # Step 4: Har example pe jao
        f_wb = w * x[i] + b  # Step 5: Prediction
        dj_dw_i = (f_wb - y[i]) * x[i]   # Step 6: w ke liye term
        dj_db_i = f_wb - y[i]            # Step 7: b ke liye term
        dj_db += dj_db_i     # Step 8: b ka sum update
        dj_dw += dj_dw_i     # Step 9: w ka sum update
    
    dj_dw = dj_dw / m        # Step 10: Average (1/m)
    dj_db = dj_db / m        # Step 11: Average (1/m)
    
    return dj_dw, dj_db      # Step 12: Dono derivatives return
```

### 🔍 Line-by-Line Breakdown

**Step 5: `f_wb = w * x[i] + b`** — Prediction nikalo: f(x) = wx + b

**Step 6: `dj_dw_i = (f_wb - y[i]) * x[i]`** — **w ke derivative ka term:** **Error × x[i]** — formula (4) se

**Step 7: `dj_db_i = f_wb - y[i]`** — **b ke derivative ka term:** **Error** (sirf, x nahi) — formula (5) se

**Step 8-9: `dj_db += dj_db_i`, `dj_dw += dj_dw_i`** — Sum mein add karo

**Step 10-11: `dj_dw = dj_dw / m`, `dj_db = dj_db / m`** — **Average nikalo** (1/m)

**Step 12: `return dj_dw, dj_db`** — **Dono derivatives** ek saath return

### 🧮 Verification (Example 1 Se Match)

| i | x | y | f_wb=100x+100 | Error | dj_dw_i = Error·x | dj_db_i = Error |
|---|---|---|---|---|---|---|
| 0 | 1 | 300 | 200 | -100 | **-100** | **-100** |
| 1 | 2 | 500 | 300 | -200 | **-400** | **-200** |

**Sums:** dj_dw = -500, dj_db = -300
**Averages (m=2):** dj_dw = **-250**, dj_db = **-150** ✓ (Example 1 se match!)

---

### 💻 Function 3: `gradient_descent` (Main Loop)

```python
def gradient_descent(x, y, w_in, b_in, alpha, num_iters, cost_function, gradient_function):
    
    J_history = []                   # Cost history store karne ke liye
    p_history = []                   # Parameters history
    b = b_in                         # b initialize
    w = w_in                         # w initialize
    
    for i in range(num_iters):       # Step 1: Loop (num_iters times)
        
        dj_dw, dj_db = gradient_function(x, y, w, b)   # Step 2: Derivatives nikalo
        
        b = b - alpha * dj_db        # Step 3: b update
        w = w - alpha * dj_dw        # Step 4: w update
        
        if i < 100000:               # Step 5: History save
            J_history.append(cost_function(x, y, w, b))
            p_history.append([w, b])
        
        if i % math.ceil(num_iters/10) == 0:   # Step 6: Har 10% pe print
            print(f"Iteration {i}: Cost {J_history[-1]}")
    
    return w, b, J_history, p_history   # Step 7: Return sab kuch
```

### 🔍 Line-by-Line Breakdown

**Step 2: `dj_dw, dj_db = gradient_function(x, y, w, b)`** — **PURANE w, b se** dono derivatives nikalo. **Ek saath** — **yehi simultaneous update ka magic hai!** ✅

**Step 3-4: `b = b - alpha * dj_db`, `w = w - alpha * dj_dw`** — Update karo

> ⚠️ **Dhoka:** Yahan b pehle update ho raha hai, phir w. **Lekin problematic nahi hai!** Kyunki **dj_dw aur dj_db PEHLE HI compute ho chuke hain** (Step 2 mein). Ab ye **purane w,b** ke basis pe hain. Bas **store karke update** kar rahe hain — **koi calculation nahi ho rahi naye w se**. So this IS simultaneous! ✅

**Step 5: `J_history.append(...)`** — Har iteration ka cost save (baad mein graph ke liye)

**Step 6: `print` har 10% pe** — Logging

**Step 7: Return** — Final w, b + history

---

### 🎮 Running The Algorithm

```python
w_init = 0
b_init = 0
iterations = 10000
tmp_alpha = 1.0e-2      # 0.01 (sahi learning rate)

w_final, b_final, J_hist, p_hist = gradient_descent(
    x_train, y_train, w_init, b_init, tmp_alpha, 
    iterations, compute_cost, compute_gradient
)
```

| **Cost badh rahi hai** | ❌ (decrease honi chahiye thi) |
| **w oscillate** | +ve → -ve → +ve (bouncing) |
| **dj_dw sign change** | Har iteration mein flip |
| **w, b ki values explode** | Bahut bade ho jaate hain |

> 🧠 **Fix:** **α kam karo!** (0.8 → 0.01)

## 🆕 Improved Visualizations (4 Runnable Code Snippets)

> 🧠 **Bhai, ye code snippets khud chala ke dekh — concept ekdum clear ho jayega!**

### 📊 Visualization 1: Cost vs Iteration (Convergence Curve)

```python
import numpy as np
import matplotlib.pyplot as plt

def compute_cost(x, y, w, b):
    m = len(x)
    total = 0
    for i in range(m):
        total += (w*x[i] + b - y[i])**2
    return total / (2*m)

def compute_gradient(x, y, w, b):
    m = len(x)
    dj_dw = dj_db = 0
    for i in range(m):
        err = w*x[i] + b - y[i]
        dj_dw += err * x[i]
        dj_db += err
    return dj_dw/m, dj_db/m

def gradient_descent(x, y, w, b, alpha, iters):
    J_hist, w_hist, b_hist = [], [], []
    for _ in range(iters):
        dw, db = compute_gradient(x, y, w, b)
        w -= alpha * dw
        b -= alpha * db
        J_hist.append(compute_cost(x, y, w, b))
        w_hist.append(w)
        b_hist.append(b)
    return w, b, J_hist, w_hist, b_hist

x = np.array([1.0, 2.0])
y = np.array([300.0, 500.0])
w, b, J, wh, bh = gradient_descent(x, y, 0, 0, 0.01, 200)

# Expected: w → 200, b → 100, cost → 0 (converged)
print(f"Final: w={w:.2f}, b={b:.2f}, cost={J[-1]:.4f}")
# Output: Final: w=200.00, b=100.00, cost=0.0000  ✓

fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].plot(J, color='steelblue')
ax[0].set_title("Cost vs Iteration (Convergence)")
ax[0].set_xlabel("Iteration"); ax[0].set_ylabel("Cost J(w,b)")
ax[0].set_yscale("log")           # Log scale for better view

ax[1].plot(wh, label="w", color='steelblue')
ax[1].plot(bh, label="b", color='orange')
ax[1].axhline(200, ls='--', color='steelblue', alpha=0.5, label='w* = 200')
ax[1].axhline(100, ls='--', color='orange', alpha=0.5, label='b* = 100')
ax[1].set_title("Parameters vs Iteration")
ax[1].set_xlabel("Iteration"); ax[1].set_ylabel("Value")
ax[1].legend()
plt.tight_layout(); plt.show()
```

**Expected Output:**
```
Final: w=200.00, b=100.00, cost=0.0000
```

**Kya Observe Karna:**
- **Left plot:** Cost **tezi se** girta hai (first 50 iters), phir flat
- **Right plot:** w, b dono **apni optimal values** (200, 100) pe converge
- **Log scale** se "tail" clearly dikhta hai

```python
def contour_path(x, y, w_hist, b_hist):
    W, B = np.meshgrid(np.linspace(150, 250, 60), np.linspace(50, 150, 60))
    J_vals = np.zeros_like(W)
    for i in range(W.shape[0]):
        for j in range(W.shape[1]):
            J_vals[i,j] = compute_cost(x, y, W[i,j], B[i,j])
    plt.figure(figsize=(9, 6))
    plt.contour(W, B, J_vals, levels=30, cmap='viridis')
    plt.plot(w_hist, b_hist, 'r.-', markersize=4, label='GD path')
    plt.plot(w_hist[0], b_hist[0], 'go', markersize=10, label='Start')
    plt.plot(w_hist[-1], b_hist[-1], 'r*', markersize=15, label='End')
    plt.xlabel('w'); plt.ylabel('b'); plt.legend()
    plt.title('Gradient Descent Path on Contour Plot')
    plt.colorbar(label='Cost J(w,b)')
    plt.show()

contour_path(x, y, wh, bh)
```

```
Gradient Descent + Linear Regression
│
├── Derivatives (∂J/∂w, ∂J/∂b)
│   ├── ∂J/∂w = (1/m)Σ(pred - actual) · x
│   │   └── Error × x (kyunki w, x se multiply hota hai)
│   ├── ∂J/∂b = (1/m)Σ(pred - actual)
│   │   └── Sirf Error (kyunki b direct add hota hai)
│   └── 1/m = average (saare examples ka)
│
├── Convexity
│   ├── Squared error cost = CONVEX (bowl)
│   ├── Sirf EK global minimum
│   └── Gradient descent ALWAYS global minimum pe jaata hai ✅
│
├── Batch Gradient Descent
│   ├── Har step: SAARE m examples dekho
│   ├── Σ from i=1 to m
│   └── (Mini-batch, Stochastic — baad mein)
│
└── Lab (3 Functions)
    ├── compute_cost: J calculate
    ├── compute_gradient: ∂J/∂w, ∂J/∂b
    │   ├── Error × x → dj_dw
    │   ├── Error → dj_db
    │   └── Average (divide by m)
    └── gradient_descent: Main loop
        ├── Derivatives nikalo (purane w,b se)
        ├── w, b update karo (simultaneous)
        ├── History save karo
        └── Repeat num_iters times
```

## 🎯 Gradient Descent + Linear Regression — Master Cheat Sheet 🔥

### 📐 Complete Algorithm

```
1. Initialize: w = 0, b = 0
2. Repeat num_iters times:
   a. Pred = w·x + b
   b. Error = Pred - y
   c. dj_dw = (1/m) Σ(Error × x)
   d. dj_db = (1/m) Σ(Error)
   e. w = w - α·dj_dw   [SIMULTANEOUS!]
   f. b = b - α·dj_db
3. Return w, b (best values)
```

### 🔑 Key Formulas

| Cheez | Formula |
|---|---|
| **Model** | f(x) = wx + b |
| **Cost** | J = (1/2m)Σ(f(x) - y)² |
| **∂J/∂w** | (1/m)Σ(f(x) - y)·x |
| **∂J/∂b** | (1/m)Σ(f(x) - y) |
| **Update** | w := w - α·∂J/∂w |

### 🧭 Derivative Intuition — 1 Line Mein

> **"Error × x = w ki galti. Error alone = b ki galti. Average nikalo (1/m)."**

### ⚠️ Divergence Signs (α Too Large)

| Symptom | Fix |
|---|---|
| Cost **badh** rahi hai | **α kam karo** |
| w **oscillate** (+ve/-ve) | **α kam karo** |
| Values **explode** | **α kam karo** |

### 🥣 Convexity — Why We're Lucky

- **Linear regression + squared error = convex**
- **Convex = bowl = only 1 global minimum**
- **Gradient descent = always global minimum** ✅
- No local minima trap! 🎉

## 🧠 Aaj Ke Confusing Parts — Simple Summary

> **Bhai, aaj ke 3 confusing cheezein thi. Unka simple breakdown:**

### 1️⃣ Derivative Formulas Kyun Aise Hain?
- **∂J/∂w mein × x** hai kyunki w, x se multiply hota hai
- **∂J/∂b mein × x nahi** kyunki b direct add hota hai
- **Bas error ka "weighted version"** hai

### 2️⃣ Calculus Skip Kar Sakte Ho?
- **Haan bilkul!** Formulas **yaad rakh** ya **code mein copy kar le**
- Course ne bola: **"Calculus skip karo, sab kaam karega"**
- **Focus: formulas ka USE** (derivatives calculate karna + update karna)

### 3️⃣ Lab Ka Code Kaise Chalta Hai?
- **compute_gradient:** Har example ka error × x, error alone — sum karo, average nikalo
- **gradient_descent:** Derivatives nikalo (purane w,b se) → w,b update → repeat
- **Simultaneous update:** Dono derivatives **pehle** compute hote hain, **phir** update

### 4️⃣ "1/2" Ka Kya Scene? (Ye Naya Doubt Tha)
- **1/2 sirf derivative ko clean rakhne ke liye** hai (Power Rule ka 2 cancel karne ke liye)
- **Minimum pe koi effect nahi** — positive constant se argmin nahi badalta
- **Convention hai** — math community ne standardize kiya
