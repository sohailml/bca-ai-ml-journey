## 🎯 Lecture 5: Classification (Supervised Learning — Part 2)

### 📌 Classification Kya Hai?
- Supervised Learning ki **dusri major type**.
- **Kya karta hai:** **Categories / Classes** predict karta hai.
- **Regression vs Classification:**
  - Regression → **Numbers** predict karta hai (infinitely many possible values)
  - Classification → **Small finite set of categories** predict karta hai

### 🩺 Example: Breast Cancer Detection

**Problem:** Doctor ke liye diagnostic tool banana jo bata sake tumor **malignant (cancerous)** hai ya **benign (safe)**.

| Label | Meaning | Value |
|---|---|---|
| **Benign** | Cancer nahi hai, safe | 0 |
| **Malignant** | Cancer hai, dangerous | 1 |

- **Input (x):** Tumor ka size
- **Output (y):** 0 ya 1 (benign ya malignant)
- **Graph:** Horizontal axis = tumor size, Vertical axis = 0 ya 1
- **Plot:** Circles = Benign, Crosses = Malignant

> 🧠 **Key Point:** Sirf **2 possible outputs** hain (0 ya 1) — isliye ye classification hai, regression nahi.

### 🔢 More Than 2 Categories Bhi Possible
- Classification mein **2 se zyada categories** bhi ho sakti hain.
- **Example:** Tumor malignant hai toh **Type 1 ya Type 2** cancer — matlab **3 possible outputs** (0, 1, 2).

> 💡 **Note:** "Output class" aur "output category" — dono ka matlab same hai, interchange mein use hote hain.

### 🎨 Categories Numbers Nahi Bhi Ho Sakti
- Categories **non-numeric** bhi ho sakti hain:
  - 🐱 Cat ya 🐶 Dog (image classification)
  - Benign ya Malignant
- Numbers bhi ho sakte hain (0, 1, 2) — **lekin sirf specific finite set**, beech ke numbers (0.5, 1.7) nahi.

### 📊 Multiple Inputs Example

**Earlier:** Sirf 1 input tha (tumor size)

**Ab:** 2 inputs —
- **x₁ = Age** (patient ki umar)
- **x₂ = Tumor size**

**Graph:** 2D plot — Age vs Tumor Size
- Circles = Benign, Crosses = Malignant
- **Algorithm kya karta hai:** Ek **boundary line** dhoondhta hai jo malignant aur benign tumors ko separate kare
- **New patient aaya** → uski age + tumor size boundary ke kis side hai, wahi decide karta hai

> 🧠 **Real-world mein:** Friends ne breast cancer detection mein **bahut zyada inputs** use kiye — tumor clump thickness, cell size uniformity, cell shape uniformity, etc.

### 📋 Recap: Supervised Learning

| Type | Kya Predict Karta Hai | Example |
|---|---|---|
| **Regression** | Numbers (infinitely many) | House price prediction |
| **Classification** | Categories (small finite set) | Benign / Malignant |

> 💡 **Key Takeaway:** Classification = **Category prediction**. Categories numbers ho sakti hain ya non-numeric — but always a **small finite set**.

## 🎯 Lecture 6: Unsupervised Learning — Clustering

### 📌 Unsupervised Learning Kya Hai?
- **Supervised mein:** Data mein **x aur y dono** hote hain (labels ke saath)
- **Unsupervised mein:** Data mein **sirf x hota hai, y (labels) nahi**
- **Algorithm ka kaam:** Data mein **structure, pattern, ya kuch interesting** dhoondhna

> 💡 Naam "unsupervised" sunke confuse mat hona — ye **bhi utna hi "super" hai** supervised jitna! 😄

### 🎯 Clustering Algorithm
- **Kya karta hai:** Unlabeled data ko **different groups (clusters)** mein divide karta hai
- **Rule:** Similar data points ek hi cluster mein aate hain

### 📰 Example 1: Google News

**Kaise kaam karta hai:**
- Google News har din **lakhs of news articles** scan karta hai
- **Similar words** wale articles ko **ek cluster** mein group karta hai

**Panda Example:**
- Headline: *"Giant panda gives birth to rare twin cubs at Japan's oldest zoo"*
- Neeche related articles dikhte hain — sabme common words: **"panda"**, **"twin"**, **"zoo"**
- **Algorithm khud figure out karta hai** — kaunse words similar articles ko indicate karte hain

> 🧠 **Magic:** Koi Google employee baithkar "panda + twin + zoo" nahi likhta. News topics **roz badalte hain** — algorithm **khud, bina supervision ke** clusters banata hai.

### 🧬 Example 2: DNA / Genetic Data

**DNA Microarray Data:**
- **Columns** = Different individuals (persons)
- **Rows** = Different genes (e.g., eye color gene, height gene, broccoli dislike gene 🥦😄)
- **Colors (red, green, gray):** Kis gene ki kitni activity hai

**Clustering ka Use:**
- Algorithm similar gene-activity wale logon ko **groups (types)** mein baant deta hai
- **Type 1, Type 2, Type 3** — automatically find ho jaate hain
- **Hum pehle se nahi batate** ki kaunse types hain — algorithm khud dhoondhta hai

### 🏢 Example 3: Market Segmentation

**Companies ke paas** customers ka huge data hota hai. Clustering se customers ko **different segments** mein divide karte hain.

**DeepLearning.AI ka real example:**
- Clustering se community ke **3 distinct groups** mile:

| Group | Motivation |
|---|---|
| **Group 1** | Knowledge seek karna, skills grow karna |
| **Group 2** | Career develop karna (promotion, new job) |
| **Group 3** | Apne field mein AI ka impact samajhna |

> 💡 **Purpose:** Har group ko better serve kar sakein.

### 📋 Clustering Summary
- **Input:** Unlabeled data (sirf x)
- **Output:** Automatically groups/clusters
- **Examples:** Google News, DNA analysis, Market segmentation

> 🧠 **Yaad Rakho:** "Next time panda dekho, toh clustering yaad karna!" 🐼

## 🎯 Lecture 7: Unsupervised Learning — Anomaly Detection & Dimensionality Reduction

### 📖 Formal Definition of Unsupervised Learning
- **Supervised:** Data mein **x + y** dono hote hain
- **Unsupervised:** Data mein **sirf x** hota hai
- **Algorithm ka kaam:** Data mein **structure / pattern / something interesting** dhoondhna

### 🔍 Types of Unsupervised Learning (is Course mein 3 Types)

#### 1️⃣ Clustering (Previous lecture)
- Similar data points ko groups mein baantna

#### 2️⃣ Anomaly Detection
- **Kya karta hai:** **Unusual events** detect karta hai
- **Main Use:** **Fraud detection** in financial systems
  - Unusual transactions = fraud ka sign
- **Other applications:** Network security, manufacturing defects, etc.

#### 3️⃣ Dimensionality Reduction
- **Kya karta hai:** Bade dataset ko **chhote dataset** mein compress karta hai
- **Magic:** **As little information loss as possible** — bina zyada information khoye
- **Analogy:** Bade room ka saara saamaan chhote box mein fit karna, lekin kuch bhi na toote

> 💡 Abhi ye terms thode confusing lagenge — **no worries**, course mein aage detail mein aayenge.

### 🧪 Quiz: Supervised vs Unsupervised

| Example | Type | Reason |
|---|---|---|
| **Spam Filtering** | ✅ Supervised | Labels hain (spam / not spam) |
| **Google News (news grouping)** | ✅ Unsupervised | Clustering algorithm, labels nahi |
| **Market Segmentation** | ✅ Unsupervised | Algorithm khud segments discover karta hai |
| **Diabetes Diagnosis** | ✅ Supervised | Labels hain (diabetes / not diabetes) — breast cancer example jaisa |

> 🧠 **Trick:** Agar data mein **labels / right answers** diye hain → **Supervised**. Agar **sirf raw data** hai aur pattern dhoondhna hai → **Unsupervised**.

### 📋 Is Course ka Structure (Unsupervised Part)
- Clustering
- Anomaly Detection
- Dimensionality Reduction

> 💡 **Key Takeaway:** Unsupervised learning = **bina labels ke data mein hidden structure dhoondhna**. 3 types: Clustering, Anomaly Detection, Dimensionality Reduction.
